## Exercise 4 -- Descriptive statistics
I should create descriptive statistics for data(mean, std, minimum value, median, quantiles, maximum)
- generate a series with 100 000 floats in normal distribution with mean of 0 and std = 100
- get descriptive statistics for this series. 
- Replace the minimum value with 5 times maximum value
- get descriptive statistics again. Did it change? if so, why?

```python
import pandas as pd
import numpy as np

g = np.random.default_rng(0)
s = pd.Series(g.normal(loc = 0, scale=100,size=100_000))
s.describe()
```

```python
modified_s = s.copy()
modified_s.loc[modified_s == modified_s.min()] = 5 * modified_s.max()
modified_s.describe()
```

**Did it change?** Yes, but only partly. `min`, `max`, `mean` and `std` changed, because they depend on every value, including the extremes: the old minimum became the new, much larger maximum (5 x old max). `median` and the quartiles (25%, 75%) are almost the same, because they depend on the order of the values, not on how extreme the outermost ones are. One replaced value out of 100 000 hardly moves them.

## Beyond task 1
demostrate than 68, 95 and 97.5% of values within 1, 2, 3 std.

```python
for k in (1, 2, 3):
    share = ((s - s.mean()).abs() <= k * s.std()).mean()
    print(f'Within {k} std: {share:.2%}')
```

## Beyond task 3
what is the mean of numbers beyond 3 std?

```python
print(f"Mean: {s.loc[(s < s.mean() - 3*s.std()) | (s > s.mean() + 3*s.std())].mean()}")
```

The mean of the values beyond 3 std is close to 0 relative to their size (they are around +-300). The distribution is symmetric, so the extreme values on the left and on the right almost cancel each other out. Their count is small (about 0.3% of the data), so the result depends on how many of them fall on each side.
