## Exercises 3 -- Counting tens digits
Create 10 random digits. And round them t0 /10. Like 10, 25, 32 should become 1, 2, 3.


```python
import pandas as pd
import numpy as np

g = np.random.default_rng(0)
s = pd.Series(g.integers(0, 101, 10))
print(s)
modifed = s/10
modifed = modifed.astype(np.int8)
print(modifed)
```

```python
modifed_2 = s.astype(str).str.get(-2).fillna('0')
print(modifed_2)
```

```python
modifed_3 = s // 10
print(modifed_3)
```

## Beyond Task 2
What is the smallest dtype for range 0 to 10000

```python
print(np.min_scalar_type(10000))
print(np.iinfo(np.uint8), np.iinfo(np.uint16), sep='\n')
s1_small = s1.astype(np.uint16)
print(s1_small.dtype, s1_small.memory_usage(index=False), 'bytes vs', s1.memory_usage(index=False), 'bytes')
```

`uint8` holds only 0-255, so it is too small. `uint16` holds 0-65535 and `int16` holds -32768 to 32767. Both fit 0-10000, so the smallest dtype is a 2-byte one, and `uint16` is the natural choice because there are no negative values.

## Beyond Task 3
new Series with 10 floating-point values between 0 and 1000. Find even numbers consedering only integer part.

```python
g2 = np.random.default_rng(0)
floats = pd.Series(g2.uniform(0, 1000, 10))
int_part = floats.astype(int)   # truncates the fractional part
even = int_part[int_part % 2 == 0]
print(pd.DataFrame({'float': floats, 'int part': int_part, 'is even': int_part % 2 == 0}))
print('Even integer parts:', even.tolist())
```
