## Exercise 5 -- Monday temperatures
Create a Series of 28 temperature readings in Celsius, representing four seven-day weeks, randomly selected from a normal distribution with a mean of 20 and a standard deviation of 5, rounded to the nearest integer.

```python
import pandas as pd
import numpy as np
from pandas import Series, DataFrame

g = np.random.default_rng(0)
days = "Sun Mon Tue Wed Thu Fri Sat".split()
days = days*4

s = Series(g.normal(20, 5, 28), index=days).round().astype(np.int8)
print(s.head())
print(s.loc['Mon'])
print(f'Monday mean: {s.loc['Mon'].mean()}')
```

## What I learned from the explanation
How to calculate statistics in a Series based on its index.

## Beyond Task 1  
What was the average weekend temperature?

```python
weekend = s.loc[['Sun', 'Sat']]
print(f'Weekend mean: {weekend.mean()}')

# .loc returns all Sundays, then all Saturdays, so % 4 gives the week number
print('Weekend mean per week:')
print(weekend.groupby(np.arange(len(weekend)) % 4).mean())
```

## Beyond Task 2
How many times is the change in temperature from the previous day greater than 2 degrees?

```python
changes = (s.diff().abs() > 2).sum()
print(f'The temperature changed by more than 2 degrees from the previous day {changes} times')
```

## Beyond Task 3
What are the two most common temperatures in our data set, and how often does each appear?

```python
counts = s.value_counts()

# keep every temperature tied with second place, which head(2) would drop
print('The most common temperatures and how often each appears:')
print(counts[counts >= counts.iloc[1]])
```
