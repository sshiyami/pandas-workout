## Exercise 1 -- Test scores
We should create exam results for every month during the year. Then find:
- average for the entire year
- average for the first and second half
- did the student improve their performance in the second half? if so, by how much?


```python
import pandas as pd
import numpy as np
g = np.random.default_rng(0)
months = 'Sep Oct Nov Dec Jan Feb Mar Apr May Jun'.split()
s = pd.Series(g.integers(70, 101, 10), index=months)
s
```

```python
print(f"Yearly average: {s.mean():.2f}")
```

```python
first_half_average = s['Sep':'Jan'].mean()
second_half_average = s['Feb':'Jun'].mean()

print(f'First half average: {first_half_average}')
print(f'Second half average: {second_half_average}')

print(f'Improvement: {second_half_average - first_half_average:.2f}')
```
