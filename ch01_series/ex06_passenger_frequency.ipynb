## Exercise 6 -- Passenger frequency
In this exercise, we begin looking at real-world data loaded from a one-column CSV file. There is a CSV file with taxi ride data. I need to compare how many rides had one passenger versus the maximum of six passengers.

```python
import pandas as pd
import numpy as np

s = pd.read_csv('../data/taxi-passenger-count.csv', header=None, names=['passengers']).squeeze()
print("First 5 numbers:")
print(s.head())
print(f"Rides with 1 passenger: {(s == 1).sum()}")
print(f"Rides with 6 passengers: {(s == 6).sum()}")
print("Number of rides for each passenger count:")
print(s.value_counts())
print("Number of rides with 1 and 6 passengers:")
print(s.value_counts()[[1, 6]])
print("Proportion of rides with 1 and 6 passengers:")
print(s.value_counts(normalize=True)[[1, 6]])
```
