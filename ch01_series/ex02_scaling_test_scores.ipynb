## Exercises 2 -- Scaling Test scores
Create scores between 40 and 60 for ten months and shift them so that the mean is 80.

```python
import pandas as pd
import numpy as np
g = np.random.default_rng(0)
months = 'Sep Oct Nov Dec Jan Feb Mar Apr May Jun'.split()
s = pd.Series(g.integers(40, 61, 10), index=months)
s
```

```python
diff = 80 - s.mean()
s = s + diff
s
```

## What I learned from the explanation
We can apply operations to a Series and the operation will be applied to each element of the Series. In this case, we added a constant value to each element of the Series to shift the values so that the mean is 80.

## Beyond 1
I should assign the grades A, B, C, D using mean and std:
- A: above mean + std
- B: from the mean up to mean + std
- C: from mean - std up to the mean
- D: below mean - std

```python
mean , std = s.mean(), s.std()
print('Months with score D: \n', s[s < mean - std].index.tolist())
print('Months with score C: \n', s[(s >= mean - std) & (s < mean)].index.tolist())
print('Months with score B: \n', s[(s >= mean) & (s < mean + std)].index.tolist())
print('Months with score A: \n', s[s >= mean + std].index.tolist())
```

## Beyond 2
Find the months that are more than 2 std above or below the mean, if there are any.

```python
mean , std = s.mean(), s.std()
print('Months more than 2 std above the mean: \n', s[s > mean + 2*std].index.tolist())
print('Months more than 2 std below the mean: \n', s[s < mean - 2*std].index.tolist())
```

## Beyond 3
How close are the mean and median to each another? What does it mean if they are close? What would it mean if they were far apart?

```python
print(f'Mean: {s.mean():.2f} \n Median: {s.median():.2f}')
```

If they are close, it means there are no anomalously high or low values (outliers). If they are far apart, the data is skewed or has outliers: the mean is pulled toward the extreme values, while the median is not.

Here the mean (80.00) is slightly higher than the median (78.30) because of the two high scores (89.8), but they are close, so the data is almost symmetric.
